```python
# -----------------------------
# 1. Basic Probability Examples
# -----------------------------
total_students = len(df)

pass_students = len(df[df["final_exam_pass"]=="Pass"])
prob_pass = pass_students/total_students

study_more_10 = len(df[df["study_hours"]>10])
prob_study_more_10 = study_more_10/total_students

attendance_80 = len(df[df["attendance"]>80])
prob_attendance_80 = attendance_80/total_students

print("\nProbability of passing exam:",prob_pass)
print("Probability of studying >10 hours:",prob_study_more_10)
print("Probability of attendance >80%:",prob_attendance_80)
```

```python
# -----------------------------
# 2. Empirical vs Theoretical
# -----------------------------
# Empirical probability of pass
empirical_pass = pass_students/total_students

# Theoretical example (coin toss)
theoretical_heads = 1/2

print("\nEmpirical Probability (Pass exam):", empirical_pass)
print("Theoretical Probability (Head in coin toss):", theoretical_heads)
```

```python
# -----------------------------
# 3. Random Variable
# X = number of students passing out of 3 randomly selected
# -----------------------------

p = prob_pass

prob_dist = {
    0: (1-p)**3,
    1: 3*p*(1-p)**2,
    2: 3*(p**2)*(1-p),
    3: p**3
}

print("\nProbability Distribution")
for k,v in prob_dist.items():
    print("X =",k," Probability =",v)

# Mean and Variance (Binomial)
mean = 3*p
variance = 3*p*(1-p)

print("\nMean:",mean)
print("Variance:",variance)
```

```python
# -----------------------------
# 4. Venn Diagram Data
# -----------------------------

study_set = set(df[df["study_hours"]>10]["final_exam_pass"])
attendance_set = set(df[df["attendance"]>80]["final_exam_pass"])

both = study_set.intersection(attendance_set)

print("\nStudents studying >10 hours:",study_set)
print("Students attendance >80:",attendance_set)
print("Students satisfying both:",both)
```

```python
# -----------------------------
# 5. Contingency Table
# -----------------------------

table = pd.crosstab(df["group_discussion"], df["final_exam_pass"])
print("\nContingency Table")
print(table)

# Joint Probability
joint = len(df[(df["group_discussion"]=="yes") & (df["final_exam_pass"]=="Pass")])/total_students

# Marginal Probability
marginal = pass_students/total_students

# Conditional Probability
yes_group = len(df[df["group_discussion"]=="yes"])
conditional = len(df[(df["group_discussion"]=="yes") & (df["final_exam_pass"]=="Pass")])/yes_group

print("\nJoint Probability (Discussion AND Pass):",joint)
print("Marginal Probability (Pass):",marginal)
print("Conditional Probability (Pass | Discussion):",conditional)
```

```python
# -----------------------------
# 6. Understanding Relationships
# -----------------------------

# Students in group discussion
discussion_yes = df[df["group_discussion"] == "yes"]

# Students who passed AND participated
both = df[(df["group_discussion"]=="yes") & (df["final_exam_pass"]=="Pass")]

# Conditional Probability P(Pass | Discussion)
conditional_probability = len(both) / len(discussion_yes)

print("Conditional Probability P(Pass | Discussion) =", conditional_probability)


# Relationship check
pass_students = len(df[df["final_exam_pass"]=="Pass"]) / total_students
discussion_students = len(df[df["group_discussion"]=="yes"]) / total_students
joint = len(both) / total_students

print("P(Pass) =", pass_students)
print("P(Discussion) =", discussion_students)
print("P(Pass AND Discussion) =", joint)

if joint == pass_students * discussion_students:
    print("Events are Independent")
else:
    print("Events are Dependent")
```

```python
# -----------------------------
# 7. Bayes Theorem
# -----------------------------

# Given probabilities
P_pass_given_high = 0.70
P_fail_given_high = 0.40
P_high = 0.60

# Total probability of pass
P_pass = (P_pass_given_high * P_high) + (P_fail_given_high * (1 - P_high))

# Bayes theorem
bayes_result = (P_pass_given_high * P_high) / P_pass

print("Probability that student passed given high attendance:")
print("P(Pass | High Attendance) =", bayes_result)
```
